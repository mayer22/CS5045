### Initial structuring of data
Using a for loop with conditions to store our state across several lines and append the state to the university in a new item in a list

The result is a list of each university with the state it is in

Once it is structured we'll create a DataFrame object with the data

```python
structured_data = []

for line in content:
    if "[edit]" in line:
        state = line
    else:
        structured_data.append((state, line))

structured_data[0:5]
```

```python
townsDF = pd.DataFrame(structured_data, columns=["State","Town"])
townsDF
```

### Clean the text
Define functions that are then applied/mapped to elements in the DataFrame object.

The input of the function can either be an entire row/column (.apply) or a single element (.map)


```python
# Check to see if an item passed into this function has "[edit]" appended to the end of it 
def clean_state(item):
    if "[edit]" in item:
        return item[ : item.find('[')] # Get the location of the start of "[edit]" and return all the text up to that
    else:
        return item

# Map the function to the DataFrame object
# This looks at each item in a Series object, including the towns (but nothing will happen to them since the text isn't in the Towns text).
townsDF = townsDF.map(clean_state) # Function creates a new copy of the DataFrame object
townsDF
```

```python
# Extract school name (inside parentheses) out of the Town index of the Series object passed into this function
# Pass in an entire row to the function (which is a Series object)
def extract_school(row):
    item = row["Town"]
    return item[item.find('(') + 1 : ] # Get the location of the parenthesis and return all the text after that

# Create a new column called "College"
# Apply takes the entire row (which is a Series object) and passes it to the function. The returned value is added to that row in the new column
townsDF["College"] = townsDF.apply(extract_school,axis=1) # axis = 1 applies function along column (i.e., Town) to each row
townsDF
```

```python
# Remove the college from the town column 
def remove_college_from_town(item):
    return item[ : item.find('(') - 1] # Find the start of the College text and take everything before it

# This time we mapped the function only to a single column/Series (Town)
townsDF["Town"] = townsDF['Town'].map(remove_college_from_town)
townsDF
```

```python
# Clean text and remove bad characters at the end 
def clean_college(item):
    item = item[ : item.find(')')]
    item = item.replace("â€“",": ")
    return item

# This time we used apply on a single column (College)
townsDF["College"] = townsDF['College'].apply(clean_college)
townsDF
```
